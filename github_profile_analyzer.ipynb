# 5. First API Call

````python
example_username = "sadiyatamanna"
url = f"https://api.github.com/users/{example_username}"

response = requests.get(url, timeout=10)

print("Status Code:", response.status_code)

if response.status_code == 200:
    data = response.json()
    print("Username:", data.get("login"))
    print("Name:", data.get("name"))
    print("Public Repositories:", data.get("public_repos"))
else:
    print("Unable to fetch profile.")
````

# 7. Build Function 1 — Fetch GitHub Profile

Now we start building the actual project.

This function:

- receives a username
- sends an API request
- returns profile data if successful
- handles common errors


````python
def get_github_profile(username):
    url = f"https://api.github.com/users/{username}"

    try:
        response = requests.get(url, timeout=10)

        if response.status_code == 200:
            return response.json()

        elif response.status_code == 404:
            print("GitHub user not found.")
            return None

        elif response.status_code == 403:
            print("GitHub API rate limit reached or access denied.")
            return None

        else:
            print("Unable to fetch profile.")
            print("Status Code:", response.status_code)
            return None

    except requests.exceptions.RequestException as error:
        print("Network error:", error)
        return None
````

# 8. Build Function 2 — Fetch Repositories

GitHub provides another endpoint for public repositories:

```text
https://api.github.com/users/{username}/repos
```

The function below requests up to 100 public repositories.


````python
def get_repositories(username):
    url = f"https://api.github.com/users/{username}/repos"

    params = {
        "per_page": 100,
        "sort": "updated"
    }

    try:
        response = requests.get(url, params=params, timeout=10)

        if response.status_code == 200:
            return response.json()

        print("Unable to fetch repositories.")
        print("Status Code:", response.status_code)
        return []

    except requests.exceptions.RequestException as error:
        print("Network error:", error)
        return []
````

### Test Repository Retrieval


````python
repos = get_repositories("sadiyatamanna")

print("Repositories received:", len(repos))

for repo in repos[:5]:
    print("-", repo.get("name"))
````

# 9. Build Function 3 — Classify the Developer Profile

We can use `if / elif / else` to create simple project rules.

````python
def classify_profile(public_repos):
    if public_repos >= 20:
        return "Highly Active Developer"
    elif public_repos >= 10:
        return "Active Developer"
    elif public_repos >= 5:
        return "Growing Developer"
    else:
        return "Beginner Developer"
````

# 10. Build Function 4 — Analyze Repositories

For every repository, we will collect:

- repository name
- primary language
- stars

We will also calculate:

- total repositories analyzed
- total stars
- average stars
- unique languages
- most frequently used language


````python
def analyze_repositories(repositories):
    repo_names = []
    languages = []
    total_stars = 0

    for repo in repositories:
        repo_names.append(repo.get("name", "Unknown"))

        language = repo.get("language")
        if language:
            languages.append(language)

        total_stars += repo.get("stargazers_count", 0)

    unique_languages = set(languages)

    language_count = {}

    for language in languages:
        if language in language_count:
            language_count[language] += 1
        else:
            language_count[language] = 1

    most_used_language = None

    if language_count:
        most_used_language = max(language_count, key=language_count.get)

    average_stars = 0

    if len(repositories) > 0:
        average_stars = total_stars / len(repositories)

    return {
        "repo_names": repo_names,
        "total_repos_analyzed": len(repositories),
        "total_stars": total_stars,
        "average_stars": average_stars,
        "unique_languages": unique_languages,
        "language_count": language_count,
        "most_used_language": most_used_language
    }
````

### Test the Analysis Function


````python
analysis = analyze_repositories(repos)

print("Repositories analyzed:", analysis["total_repos_analyzed"])
print("Total stars:", analysis["total_stars"])
print("Average stars:", round(analysis["average_stars"], 2))
print("Languages:", analysis["unique_languages"])
print("Most used language:", analysis["most_used_language"])
````

# 11. Build Function 5 — Display Profile Information


````python
def display_profile(profile):
    print("\n" + "=" * 45)
    print("           GITHUB PROFILE")
    print("=" * 45)

    print("Name            :", profile.get("name") or "Not provided")

    print("Public Repos    :", profile.get("public_repos"))
    print("Followers       :", profile.get("followers"))
    print("Following       :", profile.get("following"))
    print("Profile URL     :", profile.get("html_url"))
````

# 12. Build Function 6 — Display Repository Report


````python
def display_repository_report(repositories, analysis):
    print("\n" + "=" * 45)
    print("         REPOSITORY ANALYSIS")
    print("=" * 45)

    if not repositories:
        print("No public repositories found.")
        return

    for index, repo in enumerate(repositories[:10], start=1):
        name = repo.get("name", "Unknown")
        language = repo.get("language") or "Not specified"
        stars = repo.get("stargazers_count", 0)

        print(f"{index}. {name}")
        print("   Language:", language)
        print("   Stars   :", stars)

    print("\nSummary")
    print("-" * 45)
    print("Repositories analyzed :", analysis["total_repos_analyzed"])
    print("Total stars           :", analysis["total_stars"])
    print("Average stars         :", round(analysis["average_stars"], 2))
    print("Unique languages      :", analysis["unique_languages"])
    print("Most used language    :", analysis["most_used_language"] or "Not available")
````

# 13. Complete Project Function

The `main()` function connects all the smaller functions.

This is an example of **modular programming**:

```text
main()
 ├── get_github_profile()
 ├── get_repositories()
 ├── classify_profile()
 ├── analyze_repositories()
 ├── display_profile()
 └── display_repository_report()
```


````python
def main():
    print("=" * 50)
    print("       GITHUB PROFILE ANALYZER")
    print("=" * 50)

    username = input("Enter a GitHub username: ").strip()

    if not username:
        print("Username cannot be empty.")
        return

    print("\nFetching GitHub data...")

    profile = get_github_profile(username)

    if profile is None:
        return

    repositories = get_repositories(username)

    analysis = analyze_repositories(repositories)

    developer_level = classify_profile(profile.get("public_repos", 0))

    display_profile(profile)

    print("Profile Level   :", developer_level)

    display_repository_report(repositories, analysis)

    print("\n" + "=" * 50)
    print("Analysis completed successfully.")
    print("=" * 50)
````

````python
def display_profile(profile):
    print("\n--- GitHub Profile ---")
    print("Username       :", profile.get("login"))
    print("Name           :", profile.get("name"))
    print("Followers      :", profile.get("followers"))
    print("Following      :", profile.get("following"))
    print("Public Repos   :", profile.get("public_repos"))
    print("Bio            :", profile.get("bio"))
    print("Location       :", profile.get("location"))
    print("Company        :", profile.get("company"))
````

````python
def find_most_starred_repository(repositories):
    if not repositories:
        print("No repositories found.")
        return

    most_starred = repositories[0]

    for repo in repositories:
        if repo.get("stargazers_count", 0) > most_starred.get("stargazers_count", 0):
            most_starred = repo

    print("\n--- Most-Starred Repository ---")
    print("Repository :", most_starred.get("name"))
    print("Stars      :", most_starred.get("stargazers_count"))
````

````python
def count_total_forks(repositories):
    total_forks = 0

    for repo in repositories:
        total_forks += repo.get("forks_count", 0)

    print("\nTotal Forks :", total_forks)
````

````python
def search_repository(repositories):
    search_name = input("\nEnter repository name to search: ").strip().lower()

    found = False

    for repo in repositories:
        if repo.get("name", "").lower() == search_name:
            print("Repository found:", repo.get("name"))
            print("Description:", repo.get("description"))
            print("Stars:", repo.get("stargazers_count"))
            found = True
            break

    if not found:
        print("Repository not found.")
````

````python
def search_by_language(repositories):
    language = input("\nEnter programming language to search: ").strip().lower()

    found = False

    print("\n--- Repositories Using", language, "---")

    for repo in repositories:
        repo_language = repo.get("language")

        if repo_language and repo_language.lower() == language:
            print(repo.get("name"))
            found = True

    if not found:
        print("No repositories found using", language)
````

````python
def main():
    print("=" * 50)
    print("       GITHUB PROFILE ANALYZER")
    print("=" * 50)

    username = input("Enter a GitHub username: ").strip()

    if not username:
        print("Username cannot be empty.")
        return

    print("\nFetching GitHub data...")

    profile = get_github_profile(username)

    if profile is None:
        return

    repositories = get_repositories(username)

    analysis = analyze_repositories(repositories)

    developer_level = classify_profile(profile.get("public_repos", 0))

    display_profile(profile)

    print("Profile Level   :", developer_level)

    display_repository_report(repositories, analysis)

    find_most_starred_repository(repositories)

    count_total_forks(repositories)

    search_repository(repositories)

    search_by_language(repositories)

    print("\n" + "=" * 50)
    print("Analysis completed successfully.")
    print("=" * 50)
````

````python
def find_repositories_by_language(repositories, target_language):
    matches = []

    for repo in repositories:
        language = repo.get("language")

        if language and language.lower() == target_language.lower():
            matches.append(repo.get("name"))

    return matches


python_repos = find_repositories_by_language(repos, "Python")

print("Python repositories:", python_repos)
````
