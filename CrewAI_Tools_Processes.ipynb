```python
from crewai import LLM, Agent, Task, Crew
from crewai.tools import tool
```

```python
@tool("AI Keyword Finder")
def keyword_tool(topic: str) -> str:
    """
    Returns important keywords related to AI topic.
    """
    mapping = {
        "neural networks": "layers, weights, activation, backpropagation",
        "machine learning": "data, training, model, prediction",
        "deep learning": "CNN, RNN, transformers, embeddings"
    }
    return mapping.get(topic.lower(), "AI, models, training, inference")
```

```python
@tool("Math Solver")
def math_tool(expression: str) -> str:
    """
    Solves mathematical expressions.
    """
    try:
        return str(eval(expression))
    except:
        return "Invalid Expression"
```

```python
from google.colab import userdata
api_key=userdata.get('api_key')
```

```python
llm = LLM(
    model="gpt-4.1-nano",
    temperature=0.7,
    base_url="https://nexusapi.navigatelabs.ai",
    api_key=api_key
)
```

```python
research_agent = Agent(
    role="AI Research Analyst",
    goal="Research AI topic using keyword tool and explain clearly",
    backstory="Expert AI researcher who extracts structured knowledge.",
    tools=[keyword_tool],
    llm=llm,
    verbose=True
)

math_agent = Agent(
    role="Computation Specialist",
    goal="Perform numerical estimation if needed",
    backstory="Expert in solving calculations for AI examples.",
    tools=[math_tool],
    llm=llm,
    verbose=True
)

summarizer_agent = Agent(
    role="Summary Generator",
    goal="Create clean exam-ready summary",
    backstory="Expert academic summarizer.",
    tools=[],
    llm=llm,
    verbose=True
)
```

```python
topic = input("Enter AI Topic: ")
```

```python
research_task = Task(
    description=f"Use keyword tool and research key concepts of {topic}",
    expected_output="Research notes with keywords and explanation",
    agent=research_agent
)

math_task = Task(
    description="If any numerical example exists, calculate using math tool",
    expected_output="Mathematical reasoning output",
    agent=math_agent,
    context=[research_task]
)

summary_task = Task(
    description="Create final indexed summary from research",
    expected_output="5–7 numbered summary points",
    agent=summarizer_agent,
    context=[math_task]
)
```

```python
crew = Crew(
    tasks=[research_task, math_task, summary_task],
    process="sequential",
    verbose=True
)

result = await crew.kickoff_async()
print(result)
```

```python
manager_agent = Agent(
    role="Project Manager",
    goal="Coordinate research and summary workflow",
    backstory="Expert in managing AI research teams.",
    llm=llm,
    verbose=True,
    tracing=True
)
```

```python
crew = Crew(
    agents=[research_agent, math_agent, summarizer_agent],
    tasks=[research_task, math_task, summary_task],
    manager_agent=manager_agent,
    process="hierarchical",
    verbose=True,
    tracing=True
)

result = await crew.kickoff_async()
print(result)
```
